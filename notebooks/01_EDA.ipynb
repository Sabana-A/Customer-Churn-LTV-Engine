```python
# Select active customers
df_ltv = df_clean[df_clean["Churn"] == "No"].copy()

print("Active customers:", len(df_ltv))
```

```python
df_ltv["LTV"] = df_ltv["MonthlyCharges"] * df_ltv["tenure"]

print(df_ltv[["MonthlyCharges", "tenure", "LTV"]].head())
```

```python
# Select features for LTV prediction
X_ltv = df_ltv.drop(columns=["LTV", "Churn", "customerID", "TotalCharges"])

y_ltv = df_ltv["LTV"]

print("X_ltv shape:", X_ltv.shape)
print("y_ltv shape:", y_ltv.shape)
```

```python
# Encode categorical features
X_ltv_encoded = pd.get_dummies(X_ltv, drop_first=True)

print("X_ltv_encoded shape:", X_ltv_encoded.shape)
```

```python
from sklearn.model_selection import train_test_split

X_ltv_train, X_ltv_test, y_ltv_train, y_ltv_test = train_test_split(
    X_ltv_encoded,
    y_ltv,
    test_size=0.2,
    random_state=42
)

print("X_ltv_train shape:", X_ltv_train.shape)
print("X_ltv_test shape:", X_ltv_test.shape)
print("y_ltv_train shape:", y_ltv_train.shape)
print("y_ltv_test shape:", y_ltv_test.shape)
```

```python
from sklearn.ensemble import RandomForestRegressor

ltv_model = RandomForestRegressor(
    n_estimators=100,
    random_state=42
)

ltv_model.fit(X_ltv_train, y_ltv_train)

print("LTV model trained successfully")
```

```python
from sklearn.metrics import mean_absolute_error, r2_score

# Make predictions
ltv_pred = ltv_model.predict(X_ltv_test)

# Calculate metrics
mae = mean_absolute_error(y_ltv_test, ltv_pred)
r2 = r2_score(y_ltv_test, ltv_pred)

print("MAE:", mae)
print("R² Score:", r2)
```

```python
joblib.dump(ltv_model, "../models/ltv_model.pkl")

print("LTV model saved successfully")
```

```python
def predict_customer(churn_data, ltv_data):
    
    # Churn prediction
    churn_result = rf_model.predict(churn_data)[0]
    churn_probability = rf_model.predict_proba(churn_data)[0][1]

    # LTV prediction
    ltv_result = ltv_model.predict(ltv_data)[0]

    return {
        "churn_prediction": "Yes" if churn_result == 1 else "No",
        "churn_probability": churn_probability,
        "predicted_ltv": ltv_result
    }

print("Prediction function created successfully")
```
